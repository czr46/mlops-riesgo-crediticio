## 1. Carga del dataset

```python
df = pd.read_csv(DATA_PATH)
print(f"Filas: {df.shape[0]:,} | Columnas: {df.shape[1]}")
df.head()
```

## 2. Validación estructural básica

```python
df.info()
```

```python
# Filas duplicadas
n_dup = df.duplicated().sum()
print(f"Filas completamente duplicadas: {n_dup}")
```
